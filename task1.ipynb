```python
import json
import csv

purchases = {}

with open('data/data_task1/purchase_log.txt', encoding='utf-8') as f:
    next(f)
    for line in f:
        record = json.loads(line)
        purchases[record['user_id']] = record['category']

for user_id, category in list(purchases.items())[:2]:
    print(user_id, category)
```

```python
with open('data/data_task1/visit_log.csv', encoding='utf-8') as infile, \
     open('data/data_task1/funnel.csv', 'w', encoding='utf-8', newline='') as outfile:

    reader = csv.reader(infile)
    writer = csv.writer(outfile)

    header = next(reader)
    writer.writerow(header + ['category'])

    for row in reader:
        user_id = row[0]
        if user_id in purchases:
            writer.writerow(row + [purchases[user_id]])
```
